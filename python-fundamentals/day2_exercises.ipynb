```python
password = 123
user_entry = int(input("Enter your password:"))
print(user_entry == password)

while user_entry != password:
  print("Password is incorrect. Please try again.")

  user_entry = int(input("Enter your password:"))

  if user_entry == password:
    print("Password is correct.")
    break

```

```python
password = 123
user_entry = int(input("Enter your password:"))
print(user_entry == password)
attemps = 3

while user_entry != password and attemps >0:
  print("Password is incorrect. Please try again.")

  user_entry = int(input("Enter your password:"))

  if user_entry == password:
    print("Password is correct.")
    break
  attemps -=1
```

```python
# Create a while loop that counts from 0 to 10 and prints odd numbers to the screen.

number = 0

while number <= 10:
    if number % 2 != 0:
        print(number)
    number += 1
```

```python
secret = 44
user_number = int(input("make a guess: "))

while user_number != secret:
  user_number = int(input("Try again: "))
  if user_number < secret:
    print("close little higher!")
  elif user_number > secret:
    print("close little lower!")
  else:
    print(" You guessed it right, well done!")
```

```python
secret_number = 35
print("Lets play guess the number game.")
while True: # you can also write true
  guess = int(input("Make a guess between 1-50:"))
  if guess < secret_number:
    print("Little higher")
  elif guess > secret_number:
    print("Little lower")
  else:
    print("Correct! You guessed the number.")
    break


# you can randomize secret number.
import random
secret_number = random.randint(1,50)
print("Lets play guess the number game.")

while True: # you can also write true
  guess = int(input("Make a guess between 1-50:"))

  if guess < secret_number:
    print("Little higher")

  elif guess > secret_number:
    print("Little lower")

  else:
    print("Correct! You guessed the number.")
    break

```
