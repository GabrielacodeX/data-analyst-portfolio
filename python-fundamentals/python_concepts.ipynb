```python
print("Please enter number of hours you work each day.\nI will convert it to min.")

hours = int(input("Please enter number of hours:"))
# user entry will be converted to int

min = hours * 60

print("You work", min, "min.")
```

```python
# Write a Python program that:
# Asks the user to enter two numbers
# Adds the two numbers
# Multiply two numbers
# Divide two numbers
# Floor divide two number
# Find the remainder
# Pints the result with a message

num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))
```

```python
addition = num1 + num2
multiplication = num1 * num2
division = num1 / num2
floor_division = num1 // num2
remainder = num1 % num2

print("The sum is:", addition)
print("The product is:", multiplication)
print("The division result is:", division)
print("The floor division result is:", floor_division)
print("The remainder is:", remainder)
```
